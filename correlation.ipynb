```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("brainsize.txt", sep="\t", na_values=["NA"])
features = ["FSIQ", "VIQ", "PIQ", "Weight", "Height"]

fig, axes = plt.subplots(1, 5, figsize=(22, 4))
for ax, feature in zip(axes, features):
    sns.scatterplot(data=df, x=feature, y="MRI_Count", hue="Gender", ax=ax)
    ax.set_title(f"{feature} vs MRI_Count")
plt.tight_layout()
plt.show()
```

```python
cols = features + ["MRI_Count"]
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
for ax, gender in zip(axes, ["Female", "Male"]):
    corr = df[df["Gender"] == gender][cols].corr(method="pearson")
    sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, ax=ax)
    ax.set_title(gender)
plt.tight_layout()
plt.show()
```

Вывод:
- У мужчин размер мозга (MRI_Count) больше всего связан с PIQ (r ≈ 0.57), а с весом почти не связан (r ≈ −0.08).
- У женщин связи слабее всех, сильнее всего — с весом (r ≈ 0.45).
- Самый сильный признак по обеим группам — PIQ.
- Людей в выборке мало (по 20), так что выводы примерные, и корреляция не значит, что одно вызывает другое.
